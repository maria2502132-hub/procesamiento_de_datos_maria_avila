# Escalado
Conjunto de datos: iris (incluido en sklearn)
Este conjunto de datos es compacto y perfecto para demostrar técnicas de escalado.

```python
# Importar librerías necesarias
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargar el conjunto de datos
iris = load_iris(as_frame=True)
data = iris.data
```

```python
# Visualizar las estadísticas iniciales
print("Estadísticas iniciales:")
data.describe()
```

```python
# Aplicar MinMaxScaler
minmax_scaler = MinMaxScaler()
data_minmax = pd.DataFrame(minmax_scaler.fit_transform(data), columns=data.columns)
```

```python
# Aplicar StandardScaler
standard_scaler = StandardScaler()
data_standard = pd.DataFrame(standard_scaler.fit_transform(data), columns=data.columns)
```

### 2. Crea una gráfica de la distribución de la variable 'sepal length (cm)' antes y después del escalado.

```python

# Crear la figura
plt.figure(figsize=(12, 5))

# Distribución original
plt.subplot(1, 2, 1)
plt.hist(data['sepal length (cm)'], bins=15, edgecolor='black')
plt.title('Antes del escalado')
plt.xlabel('Longitud del sépalo (cm)')
plt.ylabel('Frecuencia')

# Distribución escalada
plt.subplot(1, 2, 2)
plt.hist(data_minmax['sepal length (cm)'], bins=15, edgecolor='black')
plt.title('Después de MinMaxScaler')
plt.xlabel('Longitud del sépalo escalada')
plt.ylabel('Frecuencia')

plt.tight_layout()
plt.show()
```

### 3. Crea una gráfica de dispersión entre 'sepal length (cm)' y 'sepal width (cm)' con los datos escalados por StandardScaler.

```python

# Crear la gráfica de dispersión
plt.figure(figsize=(8, 6))

plt.scatter(
    data_standard['sepal length (cm)'],
    data_standard['sepal width (cm)'],
    alpha=0.7
)

plt.title('Longitud vs. anchura del sépalo con StandardScaler')
plt.xlabel('Longitud del sépalo estandarizada')
plt.ylabel('Anchura del sépalo estandarizada')
plt.grid(True)
plt.show()
```

### 4. ¿Qué técnica de escalado crees que es más adecuada para este conjunto de datos? ¿Por qué?

##### La elección seria StandardScaler, ya que, centra los datos en una media de 0 y una desviación estándar de 1, es útil para algoritmos sensibles a la escala y a la distancia, y no limita los valores a un intervalo fijo; sin embargo MinMaxScaler también es una buena opción para cuando se necesita que todas las variables estén entre 0 y 1.
